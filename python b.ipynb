```python
#dictionary operations
person={'name':'bindu','age':99,'city':'england'}
print(person)
print("accessing and modify the person age:")
person["age"]=98
print("adding and removing items")
person['email']='bindu99@gmail.com'
print(person)
del person['city']
print(person)
print("All keys & values")
print(person.keys())
print(person.values())
print(person.items())
```

```python
#dict with tuple keys
location={
    (40.7128,-73.070):'New york',
    (34.0522,-119.2347):'Melbourne'}
print(location[(40.7128,-73.070)])
```

```python
#set operations
my_set={1,2,3,4,5,6}
print(my_set)
print("add and remove values")
my_set.add(7)
print(my_set)
my_set.remove(2)
print(my_set)
my_set.discard(5)
print(my_set)
print("Membership check")
print(1 in my_set)
print(10 in my_set)
```

```python
#operations
print("union")
a={1,2,3}
b={3,4,5}
print(a.union(b))
print("intersection")
print(a.intersection(b))
print("difference")
print(a.difference(b))
print("symmetric difference")
print(a.symmetric_difference(b))
```
